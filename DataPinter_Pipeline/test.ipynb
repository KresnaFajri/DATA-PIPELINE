```python
# Jalankan ini di luar Airflow dulu, langsung di Python/Jupyter
import pandas as pd
import re

df = pd.read_excel(r"/mnt/c/KRESNA/ANALYSIS/SKINCARE/DATASET SKINCARE RAW/datapinter_shopee_riset_produk_sleep_drink_20260525_100715.xlsx")  # baca file aslinya

keywords_sleep = ["nyenyak","lelap","menenangkan","tenang","sleep","relaxation",
                  "relief","stress","insomnia","susah tidur","relax","melatonin",
                  "sleep support","suplemen tidur","relaksasi","calm","calmness",
                  "rileks","tidur","stress","rileksasi","stres","calm"]

# Cek per keyword, berapa rows yang match
for kw in keywords_sleep:
    pattern = re.escape(kw.lower().replace("_", " ").strip()).replace(r"/ ", r"[/s_]+")
    count = df['Nama Produk'].str.contains(pattern, case=False, na=False, regex=True).sum()
    print(f"{kw:30s} → {count} rows")
```
